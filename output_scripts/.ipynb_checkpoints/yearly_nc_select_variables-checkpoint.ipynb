```python
import os
import glob
import numpy as np
import xarray as xr
```

```python
# --- Configuration ---
# Target a specific spinup directory based on your folder structure
data_dir = "/export/lv9/projects/dws/model_output/archived_runs/effective_fetch/spinup_01/"
out_dir = data_dir

# Use wildcard matching to grab all 12 monthly files for the year 2015
file_pattern = os.path.join(data_dir, "dws_500m.3d.2015*.nc")
monthly_files = sorted(glob.glob(file_pattern))
output_file = os.path.join(out_dir, "yearly_dws_500m.3d.2015_effective_fetch_Q.nc")

# Variables to retain. Coordinates (time, xc, yc, lonc, latc) are kept automatically.
variables_to_keep = [
    "bathymetry",
    "elev",
    "latc",
    "lonc",
    "u",
    "v",
    "ESS",
    "EIR_Ben",
    "BP1c",
    "P5c",
    "EUWIND",
    "EVWIND",
    "TauW",
    "TauC",
    "Hs_out",
    "Tz_out",
    "u_orb_out",
    "Y2c",
    "Y3c",
    "xEPS",
    "Chla",
    "netPPm2",
    "N1p",
    "N3n",
]

# Variables that are DAILY AVERAGES stamped at the END of the averaging interval.
# In each monthly file their first frame is all-NaN, and their last frame
# (labelled the 1st of the following month) holds the average of the last day.
AVERAGED_VARS = [    
    "ESS",
    "EIR_Ben",
    "BP1c",
    "P5c",
    "EUWIND",
    "EVWIND",
    "TauW",
    "TauC",
    "Hs_out",
    "Tz_out",
    "u_orb_out",
    "Y2c",
    "Y3c",
    "xEPS",
    "Chla",
    "netPPm2",
    "N1p",
    "N3n",  
]

# Instantaneous variables are stamped at the moment they describe, and each
# monthly file runs from the 1st of the month up to and including the 1st of
# the next month (i.e. one frame of overlap with the next file).

# Level index 0 is a convention slot holding the missing/fill value (-9999) for
# every variable that carries a 'level' dimension, so it must be excluded from
# the depth average. Averaging then runs over level indices 1..N-1.
DROP_FIRST_LEVEL = True

# Any remaining occurrence of this value (dry cells, land, masked columns) is
# turned into NaN before averaging so it cannot contaminate the mean.
# Set to None to disable this extra masking.
FILL_VALUE = -9999.0
FILL_TOL = 1.0  # matching tolerance, in case the value is stored imprecisely

# If True, the 1 Jan 2016 00:00 snapshot from the December file is kept, and the
# averaged variables are padded with NaN at that final timestamp (they have no
# value there). If False (default), every month is trimmed identically and the
# year ends at 31 Dec 2015 -> a clean 365-frame file with no NaN padding.
KEEP_FINAL_INSTANT = False
```

```python
# --- Pre-processing Function ---
def clean_and_average(ds):
    """
    1. Subsets variables early to save memory.
    2. Removes overlapping/invalid time steps, using a DIFFERENT rule per
       variable family:
         - instantaneous vars -> isel(time=slice(0, -1))  (drop the overlap frame)
         - daily-average vars -> isel(time=slice(1, None)) (drop the NaN frame)
       The averaged variables are then re-stamped onto the instantaneous time
       axis, so a value at time T means "mean over the 24 h starting at T",
       which lines up with the snapshot at T.
    3. Depth-averages any variable that contains a 'level' dimension.
    """
    fname = os.path.basename(ds.encoding.get("source", ""))

    # 1. Only process the variables we actually care about
    ds = ds[[v for v in variables_to_keep if v in ds]]

    # 2a. Split by how the time axis has to be treated
    timed = [v for v in ds.data_vars if "time" in ds[v].dims]
    static = [v for v in ds.data_vars if "time" not in ds[v].dims]
    inst_vars = [v for v in timed if v not in AVERAGED_VARS]
    avg_vars = [v for v in timed if v in AVERAGED_VARS]

    # 2b. Target time axis: 1st of the month .. last day of the month
    day_time = ds["time"].isel(time=slice(0, -1)).values

    # Only the December file may keep its trailing (1 Jan) snapshot
    keep_extra = KEEP_FINAL_INSTANT and "201512" in fname
    inst_time = ds["time"].values if keep_extra else day_time

    parts = []
    if static:
        parts.append(ds[static])

    if inst_vars:
        # instantaneous: drop the last frame (overlap with next month's first)
        sl = slice(0, None) if keep_extra else slice(0, -1)
        parts.append(ds[inst_vars].isel(time=sl))

    if avg_vars:
        # daily averages: drop the leading all-NaN frame instead
        da = ds[avg_vars].isel(time=slice(1, None))
        # Re-label: the frame stamped D+1 00:00 is the mean of day D
        da = da.assign_coords(time=day_time)
        for v in avg_vars:
            da[v].attrs["cell_methods"] = "time: mean (interval: 1 day)"
            da[v].attrs["time_convention"] = (
                "value at time T is the daily mean over [T, T+1 day); "
                "original file stamped this at T+1 day"
            )
        if keep_extra:
            # no average exists for the trailing snapshot -> pad with NaN
            da = da.reindex(time=inst_time)
        parts.append(da)

    # join="exact" fails loudly if the two families ever disagree in length
    ds = xr.merge(parts, join="exact", combine_attrs="drop_conflicts")

    # 3. Depth Averaging for 4D variables
    for var_name in list(ds.data_vars):
        if "level" not in ds[var_name].dims:
            continue

        da = ds[var_name]
        n_all = da.sizes["level"]

        # 3a. Drop the fill-value slot at level index 0
        if DROP_FIRST_LEVEL:
            da = da.isel(level=slice(1, None))

        # 3b. Mask any leftover sentinel values so they cannot skew the mean.
        #     NaN comparisons are False, so already-NaN cells stay NaN.
        if FILL_VALUE is not None:
            da = da.where(abs(da - FILL_VALUE) > FILL_TOL)

        # mean() skips NaN by default -> a fully masked column becomes NaN
        ds[var_name] = da.mean(dim="level", keep_attrs=True)
        ds[var_name].attrs["processing"] = "depth-averaged"
        ds[var_name].attrs["depth_average_levels"] = (
            f"level indices {1 if DROP_FIRST_LEVEL else 0}..{n_all - 1} "
            f"of {n_all}" + (" (index 0 excluded: fill value)" if DROP_FIRST_LEVEL else "")
        )

    return ds
```

```python
# --- Optional sanity check on a single file (run once) ---
# Confirms the assumptions above before you process all 12 months.
check = monthly_files[0]
with xr.open_dataset(check) as d:
    print("File:", os.path.basename(check))
    print("n time steps:", d.sizes["time"])
    print("first / last stamp:", d["time"].values[0], "/", d["time"].values[-1])
    for v in AVERAGED_VARS:
        if v in d:
            first_all_nan = bool(d[v].isel(time=0).isnull().all())
            last_all_nan = bool(d[v].isel(time=-1).isnull().all())
            print(f"  {v}: frame 0 all-NaN = {first_all_nan}, "
                  f"last frame all-NaN = {last_all_nan}")
    for v in ("temp", "salt", "elev"):
        if v in d:
            print(f"  {v}: frame 0 all-NaN = "
                  f"{bool(d[v].isel(time=0).isnull().all())}")

    print("\n-- level structure --")
    for v in variables_to_keep:
        if v in d and "level" in d[v].dims:
            col = d[v].isel(time=min(1, d.sizes["time"] - 1))
            lvl0 = col.isel(level=0)
            lvl1 = col.isel(level=1)
            print(f"  {v}: n_levels={d[v].sizes['level']}, "
                  f"level0 min/max={float(lvl0.min()):.1f}/{float(lvl0.max()):.1f}, "
                  f"level1 min/max={float(lvl1.min()):.1f}/{float(lvl1.max()):.1f}")
            print(f"      fraction of level0 equal to {FILL_VALUE}: "
                  f"{float((abs(lvl0 - FILL_VALUE) <= FILL_TOL).mean()):.3f}")
```

```python
# --- Processing ---
print(f"Found {len(monthly_files)} files. Initiating concatenation...")

ds = xr.open_mfdataset(
    monthly_files,
    preprocess=clean_and_average,
    combine="by_coords",
    data_vars="minimal",
    coords="minimal",
    compat="override",
    parallel=True,  # Leverages dask if installed for faster reading
)

if "level" in ds.coords:
    ds = ds.drop_vars("level")

# Quick verification: no duplicate or missing days
t = ds["time"].values
print("time steps:", len(t), "| first:", t[0], "| last:", t[-1])
print("duplicate stamps:", len(t) - len(np.unique(t)))

print(f"Writing combined yearly data to disk at: {output_file}")
print("This may take a few moments depending on the file sizes...")

ds.to_netcdf(output_file)
ds.close()

print("Yearly file successfully created!")
```
