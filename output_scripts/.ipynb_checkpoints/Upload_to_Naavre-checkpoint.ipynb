```python
import os
import re

base_local = "/export/lv9/projects/dws/model_output/archived_runs"
base_remote = "dws/model_output/archived_runs"

# Regex pattern: match dws_500m.3d.201501.nc through dws_500m.3d.201512.nc
pattern = re.compile(r"dws_500m\.3d\.2015(0[1-9]|1[0-2])\.nc$")

spinup_dirs = [f"spinup_{i:02d}" for i in range(1, 11)]  # spinup_01 ... spinup_10

for spinup in spinup_dirs:

    local_folder = os.path.join(base_local, spinup)
    remote_prefix = f"{base_remote}/{spinup}"

    for dirpath, dirnames, filenames in os.walk(local_folder):
        for filename in filenames:

            # Only process the desired files
            if not pattern.match(filename):
                continue

            local_path = os.path.join(dirpath, filename)

            # Skip broken symbolic links
            if os.path.islink(local_path) and not os.path.exists(local_path):
                print(f"Skipping broken symlink: {local_path}")
                continue

            # Preserve folder structure
            relative_path = os.path.relpath(local_path, local_folder)
            object_name = f"{remote_prefix}/{relative_path.replace(os.sep, '/')}"

            # Skip if remote object already exists
            if remote_exists(object_name):
                print(f"Skipping existing remote object: {object_name}")
                continue

            print(f"Uploading {local_path} -> {object_name}")
            mc.fput_object(
                bucket_name="naa-vre-waddenzee-shared",
                file_path=local_path,
                object_name=object_name
            )
```
